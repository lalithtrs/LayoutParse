# Model Evaluation & Error Typology Analysis
### SROIE Key Information Extraction Failure Mode Breakdown

This notebook classifies and investigates model failure modes:
1. **OCR Text Errors**: Character misrecognitions (e.g. `0` vs `O`, missing decimal points in totals)
2. **Boundary Truncation**: Multi-line addresses truncated by the 512-token limit or bounding box merging
3. **Semantic Ambiguity**: Multiple candidate dates (receipt printing timestamp vs transaction date)
4. **Layout Shift**: Non-standard receipt formats

```python
import json
from pathlib import Path
import pandas as pd

report_path = Path('evaluation_report.json')
if report_path.exists():
    with open(report_path) as f:
        eval_report = json.load(f)
    print('Evaluation Summary:')
    print(json.dumps(eval_report, indent=2))
else:
    print('Run `python scripts/evaluate.py` first to generate evaluation_report.json')
```
