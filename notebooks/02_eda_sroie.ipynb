# SROIE Dataset — Exploratory Data Analysis (EDA)
### ICDAR 2019 Receipt Dataset Inspection & Field Distribution

This notebook analyzes the visual and textual properties of the SROIE receipt dataset:
- Image aspect ratios and resolution distribution
- Bounding box spatial distribution
- Field value character lengths and frequencies
- Ambiguities in multi-line addresses and date formats

```python
from pathlib import Path
import json
import pandas as pd
import matplotlib.pyplot as plt
from src.data.dataset import SROIEDatasetLoader

loader = SROIEDatasetLoader()
docs = loader.load_directory('data/samples')
print(f'Loaded {len(docs)} sample receipts for EDA')
```

## 1. Inspect Field Values and Ground Truth

```python
records = []
for d in docs:
    rec = {'doc_id': d.doc_id, 'token_count': len(d.tokens)}
    rec.update(d.ground_truth)
    records.append(rec)

df = pd.DataFrame(records)
df
```

## 2. Token Length Distribution & 512 Sequence Limit

```python
print('Max tokens per document:', df['token_count'].max())
print('Mean tokens per document:', df['token_count'].mean())
print('Percentage of documents under 512 tokens:', (df['token_count'] <= 512).mean() * 100, '%')
```

## 3. BIO Tag Cardinality

```python
all_tags = [t for d in docs for t in d.bio_tags]
tag_series = pd.Series(all_tags).value_counts()
print('BIO Tag counts:\n', tag_series)
```
